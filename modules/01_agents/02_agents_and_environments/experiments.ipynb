The function `compare` runs three agents on one map: a random one, the reflex one (from `task.py` or the reference) and the reference model-based agent, which remembers the cells it has visited. For each it prints the score, the number of parcels collected and the number of steps.

```python
from pathlib import Path

import matplotlib.pyplot as plt

from introai.agents import random_agent, run_episode
from introai.agents.reference import ModelBasedAgent
from introai.world.plot import plot_map
from introai.world.warehouse import WarehouseMap

MAPS = Path.cwd().parents[2] / "data" / "maps"
parcels = WarehouseMap.from_file(MAPS / "parcels.txt")
aisles = WarehouseMap.from_file(MAPS / "aisles.txt").with_parcels(
    [(3, 1), (5, 4), (9, 21), (13, 15)]
)


def compare(warehouse, max_steps):
    """Run the three agents on one map; print and return their results."""
    agents = {
        "random": random_agent(seed=0),
        "reflex": algorithms.reflex_agent,
        "model-based": ModelBasedAgent(),
    }
    results = {}
    for name, agent in agents.items():
        result = run_episode(agent, warehouse, max_steps=max_steps)
        results[name] = result
        print(f"{name:12} score {result.score:6}  parcels {result.collected}  steps {result.steps}")
    return results


def plot_visits(warehouse, results):
    """Show the cells each of the two non-random agents visited."""
    fig, axes = plt.subplots(1, 2, figsize=(13, 4))
    for ax, name in zip(axes, ["reflex", "model-based"]):
        visited = list(dict.fromkeys(results[name].trajectory))
        plot_map(warehouse, expanded=visited, title=f"{name}: cells visited", ax=ax)
    fig.tight_layout()
```

## Experiment 1: a small map with three parcels

**Your prediction.** Which agent will do best and which worst? Will the reflex agent collect every parcel?

*My prediction:* …

```python
results = compare(parcels, max_steps=300)
plot_visits(parcels, results)
```

**Think about it.** Why did the random agent do better than the reflex one? Does that mean the random agent is rational? In the figure, blue cells are the cells the agent visited.

## Experiment 2: aisles with four parcels

A larger map and a longer limit: 1000 steps.

**Your prediction.** Will the model-based agent collect every parcel? Will its score be positive?

```python
results = compare(aisles, max_steps=1000)
plot_visits(aisles, results)
```
